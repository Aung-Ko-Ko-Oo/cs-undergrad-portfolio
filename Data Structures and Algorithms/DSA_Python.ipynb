```python
#Binary Search
def find_number_fast(sorted_numbers, target):
    low = 0
    high = len(sorted_numbers) - 1
    while low <= high:
        mid = (low + high) // 2
        if sorted_numbers[mid] == target:
            return True
        elif sorted_numbers[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return False

my_sorted_list = [1, 5, 8, 10, 20, 30]
result = find_number_fast(my_sorted_list, 10)
print("Do you find?", result) # Output: True
```

```python
#linear Search

def find_number_slow(numbers, target):
    for num in numbers:
        if num == target:
            return True
    return False

my_list = [1, 5, 8, 10, 20, 30]
result = find_number_slow(my_list, 10)
print("Do you find?", result) # Output: True
```

```python
#insertion sort
def insertion_sort(arr):
    """
    Insertion Sort implemented in Python
    Time Complexity: O(n²) - Worst case
    Space Complexity: O(1) - In-place sorting
    """
    # Start from the second element (assume the first element is already sorted)
    for i in range(1, len(arr)):
        key = arr[i]  # The current value to be inserted
        j = i - 1     # The last index of the sorted portion

        # Move elements of the sorted portion that are greater than the key
        # to one position ahead of their current position
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]  # Shift the larger value to the right
            j -= 1               # Move left to check the next element
        arr[j + 1] = key         # Insert the key into its correct position

    return arr

# Test the function
my_list = [5, 2, 4, 6, 1, 3]
print("Before sorting:", my_list)
sorted_list = insertion_sort(my_list)
print("After sorting:", sorted_list)
```

```python
def merge_sort(arr):
    """
    Merge Sort implemented in Python
    Time Complexity: O(n log n) - Best, worst, and average cases
    Space Complexity: O(n) - Requires extra space for merging
    """
    # Base case: if the list has 1 or fewer elements, it is already sorted
    if len(arr) <= 1:
        return arr

    # 1. Divide: Split the array into two halves
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    # 2. Conquer: Recursively sort both halves
    left_sorted = merge_sort(left_half)
    right_sorted = merge_sort(right_half)

    # 3. Combine: Merge the sorted halves
    return merge(left_sorted, right_sorted)

def merge(left, right):
    """
    Merge two sorted lists into a single sorted list
    """
    result = []
    i = j = 0

    # Compare elements from both lists and append the smaller one
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Append any remaining elements from either list
    result.extend(left[i:])
    result.extend(right[j:])

    return result

# Test the function
my_list = [38, 27, 43, 3, 9, 82, 10]
print("Before sorting:", my_list)
sorted_list = merge_sort(my_list)
print("After sorting:", sorted_list)
```

```python
import time
import random

def time_algorithm(sort_func, arr):
    start = time.time()
    sort_func(arr.copy())
    end = time.time()
    return end - start

# Test for an array size of 1000
test_arr = [random.randint(1, 10000) for _ in range(1000)]

insertion_time = time_algorithm(insertion_sort, test_arr)
merge_time = time_algorithm(merge_sort, test_arr)

print(f"Insertion Sort time: {insertion_time:.6f} seconds")
print(f"Merge Sort time: {merge_time:.6f} seconds")
```

```python
import matplotlib.pyplot as plt
import numpy as np

def f(n):
    return 2 * n**2 + 3 * n + 1

def g(n):
    return n**2

# Define values for n
n_values = np.arange(1, 100, 1)
f_values = [f(n) for n in n_values]
g_values = [g(n) for n in n_values]

# Find constants c₁ and c₂
c1 = 1  # Lower bound
c2 = 5  # Upper bound

lower_bound = [c1 * g(n) for n in n_values]
upper_bound = [c2 * g(n) for n in n_values]

# Plot the graph
plt.figure(figsize=(10, 6))
plt.plot(n_values, f_values, label='f(n) = 2n² + 3n + 1', linewidth=2)
plt.plot(n_values, g_values, label='g(n) = n²', linestyle='--', linewidth=2)
plt.plot(n_values, lower_bound, label=f'c₁ * g(n) = {c1}n²', linestyle=':', linewidth=2)
plt.plot(n_values, upper_bound, label=f'c₂ * g(n) = {c2}n²', linestyle=':', linewidth=2)
plt.xlabel('n (Data Size)')
plt.ylabel('f(n) (Time Complexity)')
plt.title('Visualization that f(n) = 2n² + 3n + 1 is Θ(n²)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import math
import time

def measure_time(func, n):
    """Measures the execution time of a function"""
    start = time.time()
    func(n)
    end = time.time()
    return end - start

# Functions representing different growth rates
def constant_time(n):
    return 1

def logarithmic_time(n):
    return math.log2(n)

def linear_time(n):
    return n

def linearithmic_time(n):
    return n * math.log2(n)

def quadratic_time(n):
    return n**2

def cubic_time(n):
    return n**3

def exponential_time(n):
    return 2**n

# Test with a small n
n = 10
print(f"Function values for n = {n}:")
print(f"O(1): {constant_time(n)}")
print(f"O(log n): {logarithmic_time(n):.2f}")
print(f"O(n): {linear_time(n)}")
print(f"O(n log n): {linearithmic_time(n):.2f}")
print(f"O(n²): {quadratic_time(n)}")
print(f"O(n³): {cubic_time(n)}")
print(f"O(2ⁿ): {exponential_time(n)}")

# For a larger n (some functions grow too large to compute directly)
n = 1000
print(f"\nFunction values for n = {n} (approximate):")
print(f"O(1): {constant_time(n)}")
print(f"O(log n): {logarithmic_time(n):.2f}")
print(f"O(n): {linear_time(n)}")
print(f"O(n log n): {linearithmic_time(n):.2f}")
print(f"O(n²): {quadratic_time(n):,}")
print(f"O(n³): {cubic_time(n):,}")
# exponential_time(1000) is too large to compute, so it is omitted
```
